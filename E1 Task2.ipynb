```python
# Task2
r = np.array([0.05,0.07,0.15,0.22]) #asset return
vol = np.array([0.07,0.28,0.25,0.31]) #asset volatility
rho1 = np.array([[1,0.4,0.3,0.3], [0.4,1,0.27,0.42], [0.3,0.27,1,0.5], [0.3,0.42,0.5,1]]) #asset correlation *1
#print(rho1)
rho2 = 1.3*rho1
np.fill_diagonal(rho2,1) #asset correlation *1.3
#print(rho2)
rho3 = 1.8*rho1
np.fill_diagonal(rho3,1) #asset correlation *1.8
#print(rho3) 
m=0.07  #target return

#analytical solution for tangency portfolio
def OPtR(m,r,vol,rho): #minimize variance by analytical solution
    
  cov = vol*(rho*vol).T #asset covariance
  invcov = np.linalg.inv(cov) #inverse array of asset variance
  A = np.sum(invcov)
  B = np.sum(r*invcov)
  C = r @ invcov @ r.T
  lamda = (A*m-B)/(A*C-B**2)
  gama = (C-B*m)/(A*C-B**2)
    
  w = np.dot(invcov,lamda*r+gama*1)
  
  return w
    
W1 = OPtR(m,r,vol,rho1)
W2 = OPtR(m,r,vol,rho2)
W3 = OPtR(m,r,vol,rho3)

print(W1)
print(W2)
print(W3)
```
